```python
import numpy as np
import pandas as pd
df = pd.read_csv('student_performance.csv')
df.isnull().sum()
```

```python
cleandf = df.drop_duplicates()
```

```python
cleandf['marks'] = cleandf['marks'].fillna(cleandf['marks'].median())
```

```python
cleandf.loc[cleandf['marks'] > 80, 'Performance'] = 'High'
cleandf.loc[(cleandf['marks'] < 80) & (cleandf['marks'] > 60), 'Performance'] = 'Medium'
cleandf.loc[cleandf['marks'] < 60, 'Performance'] = 'Low'
```

```python
cleandf['age'] = cleandf['age'].fillna(cleandf['age'].median())
cleandf['study_hours'] = cleandf['study_hours'].fillna(cleandf['study_hours'].median())
```

```python
cleandf = cleandf.dropna(subset=['city'])
```

```python
cleandf.reset_index(inplace = True)
```

```python
cleandf.loc[cleandf['study_hours'] >= 5, 'Study Category'] = 'Serious'
cleandf.loc[cleandf['study_hours'] < 5, 'Study Category'] = 'Casual'
```

```python
cleandf.loc[cleandf['Performance'].str.contains('High') & cleandf['Study Category'].str.contains('Serious'), 'Overall'] = 'Good'
cleandf.loc[cleandf['Performance'].str.contains('High') & cleandf['Study Category'].str.contains('Casual'), 'Overall'] = 'Brilliant'
cleandf.loc[cleandf['Performance'].str.contains('Low') & cleandf['Study Category'].str.contains('Serious'), 'Overall'] = 'Gotta focus'
cleandf.loc[cleandf['Performance'].str.contains('Low') & cleandf['Study Category'].str.contains('Casual'), 'Overall'] = 'Study Beta'
cleandf.loc[cleandf['Performance'].str.contains('Medium') & cleandf['Study Category'].str.contains('Serious'), 'Overall'] = 'Improve focus'
cleandf.loc[cleandf['Performance'].str.contains('Medium') & cleandf['Study Category'].str.contains('Casual'), 'Overall'] = 'Study more'
```
