```python
import pandas as pd
df = pd.read_csv('students.csv')
df.loc[df['Age'] >= 20, 'age_group'] = 'Adult'
df.loc[df['Age'] < 20, 'age_group'] = 'Teen'
df.loc[df['Marks'] >= 80, 'marks_band'] = 'High'
df.loc[(df['Marks'] < 80) & (df['Marks'] > 60), 'marks_band'] = 'Medium'
df.loc[df['Marks'] <= 60, 'marks_band'] = 'Low'
df.loc[(df['marks_band'] == 'High') & (df['age_group'] == 'Teen'), 'Student_profile'] = 'Brilliant!'
df.loc[(df['marks_band'] == 'High') & (df['age_group'] == 'Adult'), 'Student_profile'] = 'Good'
df.loc[(df['marks_band'] == 'Medium'), 'Student_profile'] = 'Workable'
df.loc[(df['marks_band'] == 'Low'), 'Student_profile'] = 'Loser'
```

```python
df.loc[0, 'Student_profile']
```
